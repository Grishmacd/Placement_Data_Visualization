```python
import pandas as pd
df = pd.read_csv("Placement_data_full_class.csv")
```

```python
df.head()
```

Histogram (Salary – bins = 10)

```python
plt.hist(df['salary'].dropna(), bins=10)
plt.title("Histogram of Salary")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

Bar Chart (Placement Status)

```python
status = df['status'].value_counts()

plt.bar(status.index, status.values)
plt.title("Placement Status Bar Chart")
plt.xlabel("Status")
plt.ylabel("Number of Students")
plt.show()
```

Facet Plot and Pair Plot

```python
sns.catplot(x='status', y='salary', data=df, kind='box')
plt.title("Facet Plot: Salary vs Placement Status")
plt.show()
```

Pair Plot (Academic Scores)

```python
sns.pairplot(df[['ssc_p', 'hsc_p', 'degree_p', 'status']], hue='status')
plt.show()

```

Area Chart (Placed vs Not Placed)

```python
counts = df['status'].value_counts()

plt.fill_between(counts.index, counts.values)
plt.title("Area Chart of Placement Status")
plt.xlabel("Status")
plt.ylabel("Number of Students")
plt.show()

```

Pie Chart

```python
plt.pie(counts.values, labels=counts.index, autopct='%1.1f%%')
plt.title("Placement Status Pie Chart")
plt.show()
```
